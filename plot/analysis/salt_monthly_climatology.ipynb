```python
#to calculate the anomalies - we subtract the climatology from the salinity values of each hour
salt_clim = ds.salinity 
```

```python
#for easy calculations - I will gather all the files in a dictionary, however, next time I will import the netcdf4 files in an already dictionary using os

#create dict:
may = {
    'day1' : norkyst_1,
    'day2' : norkyst_2,
    'day3' : norkyst_3,
    'day4' : norkyst_4,
    'day5' : norkyst_5,
    'day6' : norkyst_6,
    'day7' : norkyst_7,
    'day8' : norkyst_8,
    'day9' : norkyst_9,
    'day10' : norkyst_10,
    'day11' : norkyst_11,
    'day12' : norkyst_12,
    'day13' : norkyst_13,
    'day14' : norkyst_14,
    'day15' : norkyst_15,
    'day16' : norkyst_16,
    'day17' : norkyst_17,
    'day18' : norkyst_18,
    'day19' : norkyst_19,
    'day20' : norkyst_20,
    'day21' : norkyst_21,
    'day22' : norkyst_22,
    'day23' : norkyst_23,
    'day24' : norkyst_24,
    'day25' : norkyst_25,
    'day26' : norkyst_26,
    'day27' : norkyst_27,
    'day28' : norkyst_28,
    'day29' : norkyst_29,
    'day30' : norkyst_30}
```

```python
#select out salinity from the norkyst files
anomalies = {}
for day, data in may.items():
    salinity_nor = data.salinity
    anomalies[day] = salinity_nor - salt_clim
```

```python
anomalies_total = list(anomalies.values())
anomaly_ds = xr.concat(anomalies_total, dim = 'time')
```

```python
#convert into a netcdf file for easy use later
output_name = '/lustre/storeB/project/fou/hi/foccus/malene/ocean-ai/plot/analysis/anomalies_may.nc'
anomaly_ds.to_netcdf(output_name)
```
